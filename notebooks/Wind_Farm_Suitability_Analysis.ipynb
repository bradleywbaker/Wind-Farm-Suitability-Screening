# Wind Farm Suitability Screening

## Project Overview

This project evaluates potential areas for utility-scale wind farm development across the southeastern and southern United States using a GIS-based suitability screening workflow.

The analysis uses **hard minimum constraints** to exclude areas that do not meet basic development requirements. Areas that satisfy those constraints are then graded based on their relative suitability.

### Study Area

Texas, Oklahoma, Louisiana, Arkansas, Alabama, Tennessee, Mississippi, Kentucky, Georgia, Florida, South Carolina, North Carolina, Virginia, and West Virginia.

### General Workflow

**Input data → preprocessing → derived variables → hard constraints → suitability grading → analysis and visualization**

### Key Criteria

- Wind resource
- Slope and elevation
- Protected areas
- Wetlands and water
- Developed land
- Transmission infrastructure
- Roads
- Noise considerations
